# 1. บทนำ (Introduction)
การให้บริการจักรยานสาธารณะเป็นทางเลือกหนึ่งในการเดินทางในเมือง โดยจำนวนครั้งการเช่าอาจแตกต่างกันตามช่วงเวลา ฤดูกาล และสภาพอากาศ การพัฒนาตัวแบบพยากรณ์จำนวนครั้งการเช่าจึงช่วยสนับสนุนการวางแผนเตรียมบริการให้สอดคล้องกับความต้องการใช้บริการที่เปลี่ยนแปลงไป

โครงงานนี้ศึกษาข้อมูลการเช่าจักรยานรายวันของระบบ Capital Bikeshare ระหว่างปี ค.ศ. 2011–2012 จำนวน 731 วัน จาก UC Irvine Machine Learning Repository โดยมีวัตถุประสงค์เพื่อพัฒนาและเปรียบเทียบตัวแบบการถดถอยเชิงเส้น (Linear Regression Model) กับตัวแบบต้นไม้ตัดสินใจสำหรับการถดถอย (Decision Tree Regression) ในการพยากรณ์จำนวนครั้งการเช่าจักรยานรายวัน โดยใช้ภาษา Python ตั้งแต่การจัดเตรียมและทำความสะอาดข้อมูล การวิเคราะห์ข้อมูลเบื้องต้น (EDA) การคัดเลือกตัวแปร การสร้างตัวแบบ และการนำเสนอผลด้วยกราฟ

การประเมินประสิทธิภาพใช้ค่า Root Mean Squared Error (RMSE) บนชุดทดสอบ เพื่อเปรียบเทียบความคลาดเคลื่อนและเลือกตัวแบบที่เหมาะสม โดยใช้เฉพาะตัวแปรที่ทราบได้ ณ เวลาพยากรณ์ ผลการศึกษาจะนำไปสู่ข้อเสนอแนะสำหรับการประยุกต์ใช้ตัวแบบเพื่อสนับสนุนการวางแผนให้บริการจักรยานสาธารณะ

# 2. วัตถุประสงค์ของโครงงาน (Project Objectives)

เพื่อดำเนินงานให้บรรลุตาม SMART Objective ที่ตั้งไว้ โครงงานนี้จึงกำหนดวัตถุประสงค์หลักดังต่อไปนี้:

1. **เพื่อพัฒนาตัวแบบพยากรณ์ (Model Development):** สร้างและเปรียบเทียบประสิทธิภาพของ Linear regression model, Polynomial regression model, Ridge regression model, LASSO regression model และ Decision tree regression model ในการพยากรณ์จำนวนการเช่าจักรยานรายวัน
2. **เพื่อประเมินและคัดเลือกแบบจำลอง (Model Evaluation):** ประเมินประสิทธิภาพของตัวแบบบนชุดข้อมูลทดสอบ (Testing Set) เพื่อให้ได้ความคลาดเคลื่อน (RMSE) ต่ำที่สุด และมีค่า R-squared ไม่ต่ำกว่า 70% สำหรับนำไปใช้จริงได้อย่างน่าเชื่อถือ
3. **เพื่อสนับสนุนการวางแผนเชิงกลยุทธ์ (Decision Support):** นำข้อมูลเชิงลึก (Insights) จากการวิเคราะห์และผลการพยากรณ์มาจัดทำข้อเสนอแนะและแนวทางปฏิบัติในการบริหารจัดการ เพื่อสนับสนุนผู้ดูแลระบบในการเตรียมความพร้อมของจำนวนจักรยานให้สอดคล้องกับความต้องการของผู้ใช้ทั้งกลุ่มทั่วไป (Casual) และกลุ่มสมาชิก (Registered)

# 4.	Data Dictionary
| ตัวแปร | ความหมาย | ชนิดข้อมูล | รหัส / หน่วย / รายละเอียด |
|---|---|---|---|
| `instant` | ลำดับระเบียนข้อมูล | จำนวนเต็ม — ตัวระบุ | ใช้ระบุแถวข้อมูล |
| `dteday` | วันที่บันทึกข้อมูล | วันที่ | รูปแบบ `YYYY-MM-DD` |
| `season` | ฤดูกาล | หมวดหมู่ | 1 = ฤดูหนาว, 2 = ฤดูใบไม้ผลิ, 3 = ฤดูร้อน, 4 = ฤดูใบไม้ร่วง |
| `yr` | ปีที่บันทึกข้อมูล | หมวดหมู่ | 0 = ค.ศ. 2011, 1 = ค.ศ. 2012 |
| `mnth` | เดือน | หมวดหมู่ | 1 = มกราคม ถึง 12 = ธันวาคม |
| `holiday` | สถานะวันหยุด | ทวิภาค | 0 = ไม่ใช่วันหยุด, 1 = วันหยุด |
| `weekday` | วันในสัปดาห์ | หมวดหมู่ | รหัส 0–6; ตรวจสอบชื่อวันเทียบกับ `dteday` |
| `workingday` | สถานะวันทำงาน | ทวิภาค | 1 = วันทำงาน, 0 = วันเสาร์–อาทิตย์หรือวันหยุด |
| `weathersit` | ประเภทสภาพอากาศ | หมวดหมู่ | 1 = แจ่มใสหรือมีเมฆบางส่วน; 2 = มีหมอกบางหรือหมอกร่วมกับเมฆ; 3 = ฝนหรือหิมะเล็กน้อย อาจมีพายุฝนฟ้าคะนอง; 4 = ฝนหนักร่วมกับพายุหรือเม็ดน้ำแข็ง หรือหิมะร่วมกับหมอก |
| `temp` | อุณหภูมิที่ปรับสเกลแล้ว | ตัวเลขทศนิยม | ค่า normalized; ตรวจสอบสูตรก่อนแปลงกลับเป็น °C |
| `atemp` | อุณหภูมิที่ร่างกายรู้สึก ซึ่งปรับมาตราส่วนแล้ว | ตัวเลขทศนิยม | ค่า normalized; ตรวจสอบสูตรก่อนแปลงกลับเป็น °C |
| `hum` | ความชื้นสัมพัทธ์ที่ปรับสเกลแล้ว | ตัวเลขทศนิยม | ความชื้นสัมพัทธ์ (%) ÷ 100 |
| `windspeed` | ความเร็วลมที่ปรับมาตราส่วนแล้ว | ตัวเลขทศนิยม | ความเร็วลมเดิม ÷ 67 |
| `casual` | จำนวนครั้งการเช่าของผู้ใช้ทั่วไป | จำนวนเต็ม | ครั้งต่อวัน |
| `registered` | จำนวนครั้งการเช่าของสมาชิก | จำนวนเต็ม | ครั้งต่อวัน |
| `cnt` | จำนวนครั้งการเช่า (ตัวแปรเป้าหมาย) | จำนวนเต็ม | ครั้งต่อวัน; `cnt = casual + registered` |

ฤดูใบไม้ผลิ (Spring) มีการเช่าจักรยานน้อยที่สุด

* เดือน (Month): จำนวนการเช่ามีลักษณะโค้งคว่ำตามฤดูกาล โดยจะเริ่มสูงขึ้นตั้งแต่เดือนพฤษภาคม ถึงเดือนกันยายน และจะลดลงต่ำสุดในช่วงฤดูหนาวช่วงปลายปีและต้นปี

5.11 การวิเคราะห์พหุสัมพันธ์ร่วมระหว่างตัวแปรอิสระ (Multicollinearity Analysis ด้วย VIF)

เนื่องจากจากสหสัมพันธ์พบลักษณะความสัมพันธ์ที่สูงมากระหว่างอุณหภูมิ (`temp`) และอุณหภูมิที่รู้สึกจริง (`atemp`) ซึ่งอาจก่อให้เกิดปัญหา Multicollinearity ส่งผลให้ค่าสัมประสิทธิ์ของตัวแบบถดถอยเชิงเส้นไม่มีความเสถียร เราจึงทำการคำนวณค่า Variance Inflation Factor (VIF)เพื่อประเมินปัญหานี้

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

# เลือกเฉพาะตัวแปรต่อเนื่องที่เป็นฟีเจอร์หลัก
X_temp = bike[['temp', 'atemp', 'hum', 'windspeed']].copy()
# เพิ่ม intercept สำหรับคำนวณ VIF
X_temp['intercept'] = 1

vif_data = pd.DataFrame()
vif_data["Feature"] = X_temp.columns
vif_data["VIF"] = [variance_inflation_factor(X_temp.values, i) for i in range(len(X_temp.columns))]

display(vif_data[vif_data['Feature'] != 'intercept'].sort_values(by='VIF', ascending=False))
```

 สรุปผล: จะเห็นได้ว่า `temp` และ `atemp` มีค่า VIF ที่สูงมากกเกินเกณฑ์ปกติ (> 10) เนื่องจากเป็นข้อมูลที่วัดอุณหภูมิเช่นเดียวกัน ในการพัฒนาโมเดลเชิงเส้นขั้นสูง เช่น Polynomial features และ Regularization การใช้ Lasso และ Ridge จะช่วยควบคุมและลดทอนความสำคัญของตัวแปรที่มีลักษณะซ้ำซ้อนเหล่านี้ลงได้โดยอัตโนมัติ

#7.1 Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error, r2_score

# 1. สร้างและฝึกสอนโมเดลเชิงเส้น
lm_full = LinearRegression()
lm_full.fit(X_train, y_train)

# 2. ทำนายผลลัพธ์บนชุดข้อมูล Train และ Test
y_train_pred_full = lm_full.predict(X_train)
y_test_pred_full = lm_full.predict(X_test)

# 3. ประเมินประสิทธิภาพบนชุดข้อมูล Train
rmse_train_full = root_mean_squared_error(y_train, y_train_pred_full)
r2_train_full = r2_score(y_train, y_train_pred_full)

# 4. ประเมินประสิทธิภาพบนชุดข้อมูล Test
rmse_test_full = root_mean_squared_error(y_test, y_test_pred_full)
r2_test_full = r2_score(y_test, y_test_pred_full)

print("=== ประสิทธิภาพของ Linear Regression Model ===")
print(f"Train RMSE     : {rmse_train_full:.6f}")
print(f"Train R-square : {r2_train_full:.6f}")
print(f"-------------------------------------------------")
print(f"Test RMSE      : {rmse_test_full:.6f}")
print(f"Test R-square  : {r2_test_full:.6f}")
```

#7.3 Ridge Regression Model

```python
# กำหนดค่า alpha ที่ต้องการทดสอบ 6 ระดับ
alphas = [0.01, 0.1, 0.5, 1, 5, 10]

print("=== การเปรียบเทียบประสิทธิภาพของ Ridge Regression ตามค่า Alpha ===\n")

for alpha in alphas:
    # สร้างและฝึกสอนแบบจำลอง
    rdg = Ridge(alpha=alpha)
    rdg.fit(X_train_poly, y_train_poly)

    # ทำนายผล
    y_train_pred = rdg.predict(X_train_poly)
    y_test_pred = rdg.predict(X_test_poly)

    # คำนวณประสิทธิภาพสำหรับ Train set
    rmse_train = mean_squared_error(y_train_poly, y_train_pred) ** 0.5
    r2_train = r2_score(y_train_poly, y_train_pred)

    # คำนวณประสิทธิภาพสำหรับ Test set
    rmse_test = mean_squared_error(y_test_poly, y_test_pred) ** 0.5
    r2_test = r2_score(y_test_poly, y_test_pred)

    # แสดงผลลัพธ์
    print(f"Alpha: {alpha}")
    print(f"  Train RMSE : {rmse_train:.6f} | Train R-square : {r2_train:.6f}")
    print(f"  Test RMSE  : {rmse_test:.6f} | Test R-square  : {r2_test:.6f}")
    print("-" * 70)
```

#7.5 Decision Tree Regression Model

```python
# กำหนดค่า max_depth ที่ต้องการทดสอบ 6 ระดับ สำหรับ Decision Tree Regressor
depths = [2, 3, 5, 8, 10, 15]

print("=== การเปรียบเทียบประสิทธิภาพของ Decision Tree Regression ตามค่า max_depth ===\n")

for depth in depths:
    # 1. เทรนโมเดล Decision Tree Regressor ตามความลึกที่กำหนด
    dt_model = DecisionTreeRegressor(max_depth=depth, min_samples_split=10, random_state=333)
    dt_model.fit(X_train, y_train)

    # 2. ทำนายผล
    y_train_pred_dt = dt_model.predict(X_train)
    y_test_pred_dt = dt_model.predict(X_test)

    # 3. คำนวณค่า RMSE
    rmse_train_dt = mean_squared_error(y_train, y_train_pred_dt) ** 0.5
    rmse_test_dt = mean_squared_error(y_test, y_test_pred_dt) ** 0.5

    # 4. คำนวณค่า R-squared
    r2_train_dt = r2_score(y_train, y_train_pred_dt)
    r2_test_dt = r2_score(y_test, y_test_pred_dt)

    # 5. แสดงผลการเปรียบเทียบ
    print(f"Max Depth: {depth}")
    print(f"  Train RMSE : {rmse_train_dt:.4f} | Train R-square : {r2_train_dt:.4f}")
    print(f"  Test RMSE  : {rmse_test_dt:.4f}  | Test R-square  : {r2_test_dt:.4f}")
    print("-" * 70)
```

   * **Lasso Regression** มีค่า Test $R^2$ สูงที่สุดถึง 88.7% ซึ่งผ่านเกณฑ์เป้าหมายขั้นต่ำของโปรเจกต์ที่ตั้งไว้ (ไม่ต่ำกว่า 70%) อย่างมาก แสดงว่าตัวแปรที่สร้างขึ้นรวมถึงปฏิกิริยาร่วม (Interactions) สามารถพยากรณ์ความต้องการเช่าจักรยานได้อย่างมีประสิทธิภาพสูงสุดและพร้อมใช้งานจริง
   * **Ridge Regression** รองลงมาเป็นอันดับสอง มีค่า Test $R^2$ เท่ากับ 87.6%

2. **สัญญาณการเกิดปัญหา Overfitting:**
   * **Polynomial Regression:** มีค่า Train $R^2$ สูงที่สุดในกลุ่มถึง 95.4% แต่เมื่อนำไปใช้กับชุดทดสอบกลับลดลงเหลือ 84.5% ชี้ให้เห็นว่ามีความพยายามจำลองข้อมูลให้ตรงกับชุดฝึกสอนมากเกินไป แต่เมื่อประยุกต์ใช้ Regularization (Lasso/Ridge)ก็สามารถรักษาสมดุลของค่า $R^2$ บนข้อมูลใหม่ให้คงระดับสูงได้ดีมาก

3. **ข้อจำกัดของโมเดลที่เรียบง่ายเกินไป (Underfitting):**
   * **Decision Tree (Depth=5):** ให้ประสิทธิภาพต่ำที่สุดอย่างเห็นได้ชัด โดยมีค่า Test $R^2$ เพียง 52.5% เท่านั้น แสดงให้เห็นว่ารูปแบบโครงสร้างต้นไม้ไม่ยืดหยุ่นพอที่จะนำมาใช้กับแนวโน้มตามลำดับเวลาในอนาคต

# ฟังก์ชันสำหรับพยากรณ์จำนวนครั้งการเช่าจักรยาน (cnt) ด้วย Lasso Regression Model
ด้านล่างนี้คือฟังก์ชันตัวอย่างสำหรับรับค่าข้อมูลสภาพอากาศและวันเวลาใหม่ แล้วนำมาแปลงข้อมูลด้วยตัวช่วย Preprocess เดิมที่ผ่านการ `fit` แล้ว และใช้โมเดล Lasso ตัวที่ดีที่สุดในการพยากรณ์จำนวนครั้งการเช่าจักรยาน

```python
def predict_bike_rentals(new_data_df, scaler_obj, encoder_obj, poly_obj, model_obj):
    """
    ฟังก์ชันรับ DataFrame ข้อมูลใหม่ และทำนายผลลัพธ์จำนวนเช่าจักรยาน (cnt)

    Parameters:
    - new_data_df: pandas DataFrame ที่มีคอลัมน์เหมือนชุดข้อมูลตั้งต้น (ยกเว้น dteday, casual, registered, cnt)
    - scaler_obj: MinMaxScaler ที่เคยผ่านการ fit แล้ว
    - encoder_obj: OneHotEncoder ที่เคยผ่านการ fit แล้ว
    - poly_obj: PolynomialFeatures ที่เคยผ่านการ fit แล้ว
    - model_obj: แบบจำลอง Lasso ที่เทรนเสร็จแล้ว
    """
    # 1. แยกตัวแปรต่อเนื่องและตัวแปรเชิงกลุ่ม
    cat_cols = ['season', 'yr', 'mnth', 'holiday', 'weekday', 'workingday', 'weathersit']
    cont_cols = ['temp', 'hum', 'windspeed']

    # 2. ปรับสเกลข้อมูลใหม่ตามโมเดลเดิม
    scaled_features = scaler_obj.transform(new_data_df[cont_cols])
    encoded_features = encoder_obj.transform(new_data_df[cat_cols])

    # 3. รวมข้อมูลและแปลงเป็น Polynomial Features ระดับ 2
    combined_features = pd.concat([
        pd.DataFrame(scaled_features),
        pd.DataFrame(encoded_features)
    ], axis=1)

    poly_transformed = poly_obj.transform(combined_features)

    # 4. ทำนายผล
    prediction = model_obj.predict(poly_transformed)
    return prediction

# --- ตัวอย่างการใช้งานจริงด้วยการจำลองข้อมูลใหม่ขึ้นมา 1 วัน ---
# ข้อมูลจำลอง: ฤดูร้อน(3), ปี 2012(1), เดือนสิงหาคม(8), ไม่ใช่วันหยุด(0), วันพุธ(3), เป็นวันทำงาน(1), อากาศแจ่มใส(1)
# อุณหภูมิ 0.70, ความชื้น 0.55, ความเร็วลม 0.15
sample_input = pd.DataFrame([{
    'season': 3,
    'yr': 1,
    'mnth': 8,
    'holiday': 0,
    'weekday': 3,
    'workingday': 1,
    'weathersit': 1,
    'temp': 0.70,
    'hum': 0.55,
    'windspeed': 0.15
}])

# เรียกใช้งานฟังก์ชันทำนายผล
predicted_cnt = predict_bike_rentals(sample_input, scaler, encoder, poly_features, lso)

print("=== ผลการพยากรณ์สำหรับข้อมูลใหม่ ===")
display(sample_input)
print(f"\nจำนวนครั้งการเช่าจักรยานที่พยากรณ์ได้: {predicted_cnt[0]:.2f} ครั้ง/วัน")
```

จากการวิเคราะห์ความสัมพันธ์ของตัวแปร (EDA) ร่วมกับการพิจารณาค่าสัมประสิทธิ์ของตัวแบบและโครงสร้างเงื่อนไขของตัวแบบพยากรณ์ สรุปประเด็นสำคัญและข้อเสนอแนะเพื่อสนับสนุนผู้ดูแลระบบในการวางแผนจัดการจำนวนจักรยานให้สอดคล้องกับความต้องการของผู้ใช้ทั้ง 2 กลุ่มดังนี้:

#### **1. การบริหารจัดการตามปัจจัยสภาพอากาศและอุณหภูมิ**
* **ข้อมูลเชิงลึก:** อุณหภูมิ (`temp` / `atemp`) มีความสัมพันธ์เชิงบวกสูงสุดกับยอดการเช่า ($r = 0.6311$) ในขณะที่ความชื้นและสภาพอากาศที่แย่ เช่น วันที่มีฝนหรือหิมะตกเบาบาง จะทำให้ปริมาณความต้องการใช้งานลดลงอย่างเห็นได้ชัด
* **ข้อเสนอแนะเชิงกลยุทธ์:** ผู้ดูแลระบบควรใช้ระบบพยากรณ์อากาศล่วงหน้าประกอบกับการใช้ตัวแบบพยากรณ์จำนวนครั้งการเช่าเพื่อวางแผนพนักงานและโควตาจักรยานรายวัน:
  * **วันที่มีอากาศอบอุ่นและแจ่มใส:** เตรียมสำรองจักรยานเพิ่มขึ้น 15-20% ที่หน้าสถานีบริการหลักเพื่อรองรับการใช้งานที่หนาแน่น
  * **วันที่มีสภาพอากาศปิดหรือฝนตกหนัก:** ปรับลดปริมาณจักรยานที่สแตนด์บายลงเพื่อลดการเสื่อมสภาพ และสลับรอบเวลาในการซ่อมบำรุงเชิงป้องกัน (Preventive Maintenance) รถจักรยานในช่วงเวลานี้แทน

  * จำนวนครั้งการเช่าของกลุ่มผู้ใช้ทั่วไป (`casual`) จะเติบโตอย่างรวดเร็วเฉพาะในช่วงวันหยุดสุดสัปดาห์ และวันหยุดเทศกาล
* **ข้อเสนอแนะเชิงกลยุทธ์:**
  * **วันทำงานปกติ (Working Days):** เน้นกระจายและจัดสรรจักรยานให้เพียงพอที่สถานีจุดเชื่อมต่อระบบขนส่งสาธารณะ ย่านธุรกิจ และสำนักงาน เพื่ออำนวยความสะดวกแก่ผู้เดินทางกลุ่มสมาชิกที่ต้องการความเร็วและความสม่ำเสมอ
  * **วันหยุด (Weekends/Holidays):** ดำเนินการโยกย้ายและเพิ่มโควตาสำรองจักรยานไปยังสถานีรอบนอก สวนสาธารณะ และแหล่งท่องเที่ยวหลัก พร้อมทั้งเสนอโปรโมชันชั่วคราวเพื่อกระตุ้นผู้ใช้ประเภท `casual` ให้ทดลองสมัครสมาชิกรายปีเพื่อสร้างฐานรายได้ที่สม่ำเสมอและยั่งยืนให้แก่องค์กร

เมื่อเราประยุกต์ใช้ Regularization (Lasso และ Ridge) โมเดลทั้งสองสามารถควบคุมค่าน้ำหนัก (Weights) ของสัมประสิทธิ์ตัวแปรไม่ให้สูงผิดปกติ ทำให้โมเดลมีความยืดหยุ่นสูงขึ้นและพยากรณ์ข้อมูลอนาคตได้ดีขึ้นอย่างมาก (Lasso RMSE ลดลงเหลือ 674.10, Ridge RMSE ลดลงเหลือ 706.35)
นอกจากนั้น Lasso ที่มีจุดเด่นในการทำ Feature Selection อัตโนมัติ โดยการตัดฟีเจอร์ที่ไม่มีนัยสำคัญหรือซ้ำซ้อนให้มีสัมประสิทธิ์เป็น 0 ส่งผลให้แบบจำลองมีความเรียบง่าย (Simplicity) และมีความคลาดเคลื่อนต่ำที่สุดเมื่อทดสอบเชิงเวลา

3. **ข้อจำกัดของ Linear Regression และ Decision Tree**
Linear Regression ทั่วไป มีข้อจำกัดในการดึงปฏิกิริยาร่วม (Interaction terms) ของสภาพอากาศและฤดูกาล ส่งผลให้เมื่อนำไปคาดการณ์อนาคตตามลำดับเวลามีความผิดพลาดค่อนข้างสูง (Testing $R^2$ = 65.34% และ RMSE = 1103.67)
Decision Tree Regression (Max Depth = 5) แม้จะสามารถอธิบายกฎเกณฑ์เงื่อนไขการตัดตัวแปรได้ง่าย แต่มีข้อจำกัดด้านเสถียรภาพและเกิดการ Overfitting ได้ง่ายมาก ส่งผลให้ได้ประสิทธิภาพต่ำที่สุดในการทดลองนี้ โดยมี Testing $R^2$ เพียง 52.51% และมี RMSE สูงที่สุดถึง 1291.88

# 1. บทนำ (Introduction)
การให้บริการจักรยานสาธารณะเป็นทางเลือกหนึ่งในการเดินทางในเมือง โดยจำนวนครั้งการเช่าอาจแตกต่างกันตามช่วงเวลา ฤดูกาล และสภาพอากาศ การพัฒนาตัวแบบพยากรณ์จำนวนครั้งการเช่าจึงช่วยสนับสนุนการวางแผนเตรียมบริการให้สอดคล้องกับความต้องการใช้บริการที่เปลี่ยนแปลงไป

โครงงานนี้ศึกษาข้อมูลการเช่าจักรยานรายวันของระบบ Capital Bikeshare ระหว่างปี ค.ศ. 2011–2012 จำนวน 731 วัน จาก UC Irvine Machine Learning Repository โดยมีวัตถุประสงค์เพื่อพัฒนาและเปรียบเทียบตัวแบบการถดถอยเชิงเส้น (Linear Regression Model) กับตัวแบบต้นไม้ตัดสินใจสำหรับการถดถอย (Decision Tree Regression) ในการพยากรณ์จำนวนครั้งการเช่าจักรยานรายวัน โดยใช้ภาษา Python ตั้งแต่การจัดเตรียมและทำความสะอาดข้อมูล การวิเคราะห์ข้อมูลเบื้องต้น (EDA) การคัดเลือกตัวแปร การสร้างตัวแบบ และการนำเสนอผลด้วยกราฟ

การประเมินประสิทธิภาพใช้ค่า Root Mean Squared Error (RMSE) บนชุดทดสอบ เพื่อเปรียบเทียบความคลาดเคลื่อนและเลือกตัวแบบที่เหมาะสม โดยใช้เฉพาะตัวแปรที่ทราบได้ ณ เวลาพยากรณ์ ผลการศึกษาจะนำไปสู่ข้อเสนอแนะสำหรับการประยุกต์ใช้ตัวแบบเพื่อสนับสนุนการวางแผนให้บริการจักรยานสาธารณะ

# 2. วัตถุประสงค์ของโครงงาน (Project Objectives)

เพื่อดำเนินงานให้บรรลุตาม SMART Objective ที่ตั้งไว้ โครงงานนี้จึงกำหนดวัตถุประสงค์หลักดังต่อไปนี้:

1. **เพื่อพัฒนาตัวแบบพยากรณ์ (Model Development):** สร้างและเปรียบเทียบประสิทธิภาพของ Linear regression model, Polynomial regression model, Ridge regression model, LASSO model และ Decision tree regression model ในการพยากรณ์จำนวนการเช่าจักรยานรายวัน
2. **เพื่อประเมินและคัดเลือกแบบจำลอง (Model Evaluation):** ประเมินประสิทธิภาพของตัวแบบบนชุดข้อมูลทดสอบ (Testing Set) เพื่อให้ได้ความคลาดเคลื่อน (RMSE) ต่ำที่สุด และมีค่า R-squared ไม่ต่ำกว่า 70% และมีความคลาดเคลื่อน (RMSE) ต่ำที่สุด สำหรับนำไปใช้จริงได้อย่างน่าเชื่อถือ
3. **เพื่อสนับสนุนการวางแผนเชิงกลยุทธ์ (Decision Support):** นำข้อมูลเชิงลึก (Insights) จากการวิเคราะห์และผลการพยากรณ์มาจัดทำข้อเสนอแนะและแนวทางปฏิบัติในการบริหารจัดการ เพื่อสนับสนุนผู้ดูแลระบบในการเตรียมความพร้อมของจำนวนจักรยานให้สอดคล้องกับความต้องการของผู้ใช้ทั้งกลุ่มทั่วไป (Casual) และกลุ่มสมาชิก (Registered)

# 4.	Data Dictionary
| ตัวแปร | ความหมาย | ชนิดข้อมูล | รหัส / หน่วย / รายละเอียด |
|---|---|---|---|
| `instant` | ลำดับระเบียนข้อมูล | จำนวนเต็ม — ตัวระบุ | ใช้ระบุแถวข้อมูล |
| `dteday` | วันที่บันทึกข้อมูล | วันที่ | รูปแบบ `YYYY-MM-DD` |
| `season` | ฤดูกาล | หมวดหมู่ | 1 = ฤดูหนาว, 2 = ฤดูใบไม้ผลิ, 3 = ฤดูร้อน, 4 = ฤดูใบไม้ร่วง |
| `yr` | ปีที่บันทึกข้อมูล | หมวดหมู่ | 0 = ค.ศ. 2011, 1 = ค.ศ. 2012 |
| `mnth` | เดือน | หมวดหมู่ | 1 = มกราคม ถึง 12 = ธันวาคม |
| `holiday` | สถานะวันหยุด | ทวิภาค | 0 = ไม่ใช่วันหยุด, 1 = วันหยุด |
| `weekday` | วันในสัปดาห์ | หมวดหมู่ | รหัส 0–6; ตรวจสอบชื่อวันเทียบกับ `dteday` |
| `workingday` | สถานะวันทำงาน | ทวิภาค | 1 = วันทำงาน, 0 = วันเสาร์–อาทิตย์หรือวันหยุด |
| `weathersit` | ประเภทสภาพอากาศ | หมวดหมู่ | 1 = แจ่มใสหรือมีเมฆบางส่วน; 2 = มีหมอกบางหรือหมอกร่วมกับเมฆ; 3 = ฝนหรือหิมะเล็กน้อย อาจมีพายุฝนฟ้าคะนอง; 4 = ฝนหนักร่วมกับพายุหรือเม็ดน้ำแข็ง หรือหิมะร่วมกับหมอก |
| `temp` | อุณหภูมิที่ปรับสเกลแล้ว | ตัวเลขทศนิยม | ค่า normalized; ตรวจสอบสูตรก่อนแปลงกลับเป็น °C |
| `atemp` | อุณหภูมิที่ร่างกายรู้สึก ซึ่งปรับมาตราส่วนแล้ว | ตัวเลขทศนิยม | ค่า normalized; ตรวจสอบสูตรก่อนแปลงกลับเป็น °C |
| `hum` | ความชื้นสัมพัทธ์ที่ปรับสเกลแล้ว | ตัวเลขทศนิยม | ความชื้นสัมพัทธ์ (%) ÷ 100 |
| `windspeed` | ความเร็วลมที่ปรับมาตราส่วนแล้ว | ตัวเลขทศนิยม | ความเร็วลมเดิม ÷ 67 |
| `casual` | จำนวนครั้งการเช่าของผู้ใช้ทั่วไป | จำนวนเต็ม | ครั้งต่อวัน |
| `registered` | จำนวนครั้งการเช่าของสมาชิก | จำนวนเต็ม | ครั้งต่อวัน |
| `cnt` | จำนวนครั้งการเช่า (ตัวแปรเป้าหมาย) | จำนวนเต็ม | ครั้งต่อวัน; `cnt = casual + registered` |

5.7 Bivariate Analysis

```python
plt.figure(figsize = [16,4])

# Subplot 1: Year
plt.subplot(131)
sns.barplot(x='yr', y='cnt', data = bike, hue='yr', palette='pastel', legend=False)
plt.xticks(ticks=[0, 1], labels=['2011', '2012'])
plt.xlabel('Year')
plt.ylabel('Total Rentals (cnt)')

# Subplot 2: Holiday
plt.subplot(132)
sns.barplot(x='holiday', y='cnt', data = bike, hue='holiday', palette='Set2', legend=False)
plt.xticks(ticks=[0, 1], labels=['No', 'Yes'])
plt.xlabel('Holiday')
plt.ylabel('Total Rentals (cnt)')

# Subplot 3: Workingday
plt.subplot(133)
sns.barplot(x='workingday', y='cnt', data = bike, hue='workingday', palette='Accent', legend=False)
plt.xticks(ticks=[0, 1], labels=['No', 'Yes'])
plt.xlabel('Working Day')
plt.ylabel('Total Rentals (cnt)')

plt.tight_layout()
plt.show()
```

- ปี (Year): จำนวนการเช่าจักรยานเพิ่มขึ้นจากปี 2011 เป็นปี 2012

- วันหยุด (Holiday): จำนวนการเช่าจักรยานในวันที่ไม่ใช่วันหยุด (No) มากกว่าในวันหยุด (Yes)

- วันทำงาน (Workingday): จำนวนการเช่าจักรยานในวันทำงาน (Yes) และวันที่ไม่ใช่วันทำงาน (No) แตกต่างกันเพียงเล็กน้อย

```python
plt.figure(figsize=(20, 12))

# Subplot 1: Season
plt.subplot(2,3,1)
sns.boxplot(x = 'season', y = 'cnt', data = bike, hue = 'season', palette = 'pastel', legend = False)
plt.xticks(ticks=[0, 1, 2, 3], labels=['Winter', 'Spring', 'Summer', 'Fall'])
plt.xlabel('Season')
plt.ylabel('Total Rentals (cnt)')

# Subplot 2: Month
plt.subplot(2,3,2)
sns.boxplot(x = 'mnth', y = 'cnt', data = bike, hue = 'mnth', palette = 'Set3', legend = False)
plt.xticks(ticks=list(range(12)), labels=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.xlabel('Month')
plt.ylabel('Total Rentals (cnt)')

# Subplot 3: Weather Situation
plt.subplot(2,3,3)
sns.boxplot(x = 'weathersit', y = 'cnt', data = bike, hue = 'weathersit', palette = 'Set2', legend = False)
plt.xticks(ticks=[0, 1, 2], labels=['Clear', 'Mist/Cloudy', 'Light Rain/Snow'])
plt.xlabel('Weather Situation')
plt.ylabel('Total Rentals (cnt)')

# Subplot 4: Holiday
plt.subplot(2,3,4)
sns.boxplot(x = 'holiday', y = 'cnt', data = bike, hue = 'holiday', palette = 'Accent', legend = False)
plt.xticks(ticks=[0, 1], labels=['No', 'Yes'])
plt.xlabel('Holiday')
plt.ylabel('Total Rentals (cnt)')

# Subplot 5: Weekday
plt.subplot(2,3,5)
sns.boxplot(x = 'weekday', y = 'cnt', data = bike, hue = 'weekday', palette = 'hls', legend = False)
plt.xticks(ticks=list(range(7)), labels=['Sun', 'Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat'])
plt.xlabel('Weekday')
plt.ylabel('Total Rentals (cnt)')

# Subplot 6: Working Day
plt.subplot(2,3,6)
sns.boxplot(x = 'workingday', y = 'cnt', data = bike, hue = 'workingday', palette = 'coolwarm', legend = False)
plt.xticks(ticks=[0, 1], labels=['No', 'Yes'])
plt.xlabel('Working Day')
plt.ylabel('Total Rentals (cnt)')

plt.tight_layout()
plt.show()
```

* ฤดูกาล (Season): ช่วงฤดูร้อน (Summer) และฤดูใบไม้ร่วง (Fall) มีค่ากลางและจำนวนการเช่าจักรยานที่สูงที่สุดอย่างชัดเจน
ฤดูใบไม้ผลิ (Spring) มีการเช่าจักรยานน้อยที่สุด

* เดือน (Month): จำนวนการเช่ามีลักษณะโค้งคว่ำตามฤดูกาล โดยจะเริ่มสูงขึ้นตั้งแต่เดือนพฤษภาคม ถึงเดือนกันยายน และจะลดลงต่ำสุดในช่วงฤดูหนาวช่วงปลายปีและต้นปี

* สภาพอากาศ (Weathersit): วันที่อากาศแจ่มใส (Clear) มีปริมาณการเช่าสูงสุด รองลงมาคือวันที่มีหมอกหรือเมฆมาก (Mist/Cloudy)

```python
# การปรับสเกลตัวแปรต่อเนื่องด้วย MinMaxScaler
scaler = MinMaxScaler()
encoder = OneHotEncoder(sparse_output=False)
```

```python
# การรวมข้อมูลตัวแปรต่อเนื่องที่ปรับสเกลแล้วกับข้อมูลตัวแปรเชิงกลุ่มที่เข้ารหัสแล้ว
x_combined = pd.concat([pd.DataFrame(x_scaled),pd.DataFrame(x_encoded)],axis=1)
y_train = df['cnt']
```

```python
# การแบ่งชุดข้อมูลเป็นชุดฝึกสอนและชุดทดสอบ (Training and Testing sets) ในอัตราส่วน 80:20
X_train, X_test, y_train, y_test = train_test_split(x_combined, y_train, test_size = 0.2, random_state = 42)
```

#7.1 Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error, r2_score

# 1. สร้างและฝึกสอนโมเดลเชิงเส้น
lm_full = LinearRegression()
lm_full.fit(X_train, y_train)

# 2. ทำนายผลลัพธ์บนชุดข้อมูล Train และ Test
y_train_pred_full = lm_full.predict(X_train)
y_test_pred_full = lm_full.predict(X_test)

# 3. ประเมินประสิทธิภาพบนชุดข้อมูล Train
rmse_train_full = root_mean_squared_error(y_train, y_train_pred_full)
r2_train_full = r2_score(y_train, y_train_pred_full)

# 4. ประเมินประสิทธิภาพบนชุดข้อมูล Test
rmse_test_full = root_mean_squared_error(y_test, y_test_pred_full)
r2_test_full = r2_score(y_test, y_test_pred_full)

print("=== ประสิทธิภาพของ Linear Regression Model ===")
print(f"Train RMSE     : {rmse_train_full:.6f}")
print(f"Train R-square : {r2_train_full:.6f}")
print(f"-------------------------------------------------")
print(f"Test RMSE      : {rmse_test_full:.6f}")
print(f"Test R-square  : {r2_test_full:.6f}")
```

```python
#วาดกราฟเปรียบเทียบระหว่างค่าจริง (y_test) และค่าพยากรณ์ (y_test_pred_full)
plt.figure(figsize=(5, 4))
plt.scatter(y_test, y_test_pred_full, alpha=0.6, color='royalblue', label='Predicted vs Actual')

# วาดเส้นทแยงมุมแสดง Perfect Prediction
max_val = max(y_test.max(), y_test_pred_full.max())
min_val = min(y_test.min(), y_test_pred_full.min())
plt.plot([min_val, max_val], [min_val, max_val], color='firebrick', linestyle='--', linewidth=2, label='Perfect Prediction')

plt.title('Actual vs Predicted (Linear model on testing set)', fontsize=12)
plt.xlabel('y_test (Actual)', fontsize=11)
plt.ylabel('y_pred_linear (Predicted)', fontsize=11)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# เตรียมข้อมูลสำหรับการพล็อต
models = [
    "Linear\nRegression",
    "Polynomial\n(Deg 2)",
    "Ridge\n(Alpha=0.5)",
    "Lasso\n(Alpha=0.5)",
    "Decision Tree\n(Depth=5)"
]

train_rmse = [749.6890, 411.8653, 468.5563, 461.6057, 668.9518]
test_rmse = [798.2933, 788.3942, 706.3502, 674.1049, 852.4326]

train_r2 = [0.8467, 0.9537, 0.9401, 0.9419, 0.8779]
test_r2 = [0.8411, 0.8450, 0.8756, 0.8867, 0.8188]

x = np.arange(len(models))
width = 0.35

# สร้างพื้นที่วาดกราฟแบบ 2 ซับพล็อต (ซ้าย: RMSE, ขวา: R-square)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# 1. กราฟเปรียบเทียบ RMSE (ค่ายิ่งน้อยยิ่งดี)
rects1_1 = ax1.bar(x - width/2, train_rmse, width, label='Train RMSE', color='#1f77b4', alpha=0.85)
rects1_2 = ax1.bar(x + width/2, test_rmse, width, label='Test RMSE', color='#ff7f0e', alpha=0.85)
ax1.set_ylabel('RMSE Value (Lower is Better)', fontsize=12)
ax1.set_title('Model Evaluation: RMSE Comparison', fontsize=14, fontweight='bold', pad=15)
ax1.set_xticks(x)
ax1.set_xticklabels(models, fontsize=10)
ax1.legend(fontsize=11)
ax1.grid(True, linestyle='--', alpha=0.5)

# เพิ่มป้ายแสดงค่าตัวเลขบนแท่งกราฟของ RMSE
for rect in rects1_1 + rects1_2:
    height = rect.get_height()
    ax1.annotate(f'{height:.1f}',
                 xy=(rect.get_x() + rect.get_width() / 2, height),
                 xytext=(0, 3),  # offset 3 points vertically
                 textcoords="offset points",
                 ha='center', va='bottom', fontsize=9)

# 2. กราฟเปรียบเทียบ R-squared (ค่ายิ่งใกล้ 1.0 ยิ่งดี)
rects2_1 = ax2.bar(x - width/2, train_r2, width, label='Train R-squared', color='#2ca02c', alpha=0.85)
rects2_2 = ax2.bar(x + width/2, test_r2, width, label='Test R-squared', color='#d62728', alpha=0.85)
ax2.set_ylabel('R-squared (Higher is Better)', fontsize=12)
ax2.set_title('Model Evaluation: R-squared Comparison', fontsize=14, fontweight='bold', pad=15)
ax2.set_xticks(x)
ax2.set_xticklabels(models, fontsize=10)
ax2.set_ylim(0, 1.1)  # ปรับขอบเขตแกน Y ให้เห็นชัดเจน
ax2.legend(fontsize=11)
ax2.grid(True, linestyle='--', alpha=0.5)

# เพิ่มป้ายแสดงค่าตัวเลขบนแท่งกราฟของ R-squared
for rect in rects2_1 + rects2_2:
    height = rect.get_height()
    ax2.annotate(f'{height:.3f}',
                 xy=(rect.get_x() + rect.get_width() / 2, height),
                 xytext=(0, 3),
                 textcoords="offset points",
                 ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

### **วัตถุประสงค์ที่ 3: การสนับสนุนการวางแผนเชิงกลยุทธ์ (Decision Support & Strategic Insights)**
จากการวิเคราะห์ความสัมพันธ์ของตัวแปร (EDA) ร่วมกับการพิจารณาค่าสัมประสิทธิ์ของตัวแบบและโครงสร้างเงื่อนไขของตัวแบบพยากรณ์ สรุปประเด็นสำคัญและข้อเสนอแนะเพื่อสนับสนุนผู้ดูแลระบบในการวางแผนจัดการจำนวนจักรยานให้สอดคล้องกับความต้องการของผู้ใช้ทั้ง 2 กลุ่มดังนี้:

#### **1. การบริหารจัดการตามปัจจัยสภาพอากาศและอุณหภูมิ**
* **ข้อมูลเชิงลึก:** อุณหภูมิ (`temp` / `atemp`) มีความสัมพันธ์เชิงบวกสูงสุดกับยอดการเช่า ($r = 0.6311$) ในขณะที่ความชื้นและสภาพอากาศที่แย่ เช่น วันที่มีฝนหรือหิมะตกเบาบาง จะทำให้ปริมาณความต้องการใช้งานลดลงอย่างเห็นได้ชัด
* **ข้อเสนอแนะเชิงกลยุทธ์:** ผู้ดูแลระบบควรใช้ระบบพยากรณ์อากาศล่วงหน้าประกอบกับการใช้ตัวแบบพยากรณ์จำนวนครั้งการเช่าเพื่อวางแผนพนักงานและโควตาจักรยานรายวัน:
  * **วันที่มีอากาศอบอุ่นและแจ่มใส:** เตรียมสำรองจักรยานเพิ่มขึ้น 15-20% ที่หน้าสถานีบริการหลักเพื่อรองรับการใช้งานที่หนาแน่น
  * **วันที่มีสภาพอากาศปิดหรือฝนตกหนัก:** ปรับลดปริมาณจักรยานที่สแตนด์บายลงเพื่อลดการเสื่อมสภาพ และสลับรอบเวลาในการซ่อมบำรุงเชิงป้องกัน (Preventive Maintenance) รถจักรยานในช่วงเวลานี้แทน

  * จำนวนครั้งการเช่าของกลุ่มผู้ใช้ทั่วไป (`casual`) จะเติบโตอย่างรวดเร็วเฉพาะในช่วงวันหยุดสุดสัปดาห์ และวันหยุดเทศกาล

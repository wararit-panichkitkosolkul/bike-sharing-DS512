# 1. บทนำ (Introduction)

# 2. วัตถุประสงค์ (SMART Objectives)

วิเคราะห์ข้อมูลการเช่าจักรยานรายวันของระบบ Capital Bikeshare ระหว่างปี ค.ศ. 2011–2012 จำนวน 731 วัน ด้วย Python และ Tableau เพื่อระบุรูปแบบการใช้บริการตามช่วงเวลา ประเภทผู้ใช้ และสภาพอากาศ โดยจัดทำแดชบอร์ดที่ตอบคำถามการวิเคราะห์ครบทั้ง 14 ข้อ และสรุปข้อมูลเชิงลึกพร้อมข้อเสนอแนะที่มีหลักฐานสนับสนุนอย่างน้อย 3 ประเด็น เพื่อประกอบการวางแผนให้บริการและกิจกรรมส่งเสริมการใช้จักรยาน โดยดำเนินการให้แล้วเสร็จภายใน 4 สัปดาห์

| องค์ประกอบ SMART | รายละเอียด |
|---|---|
| **S — Specific** เฉพาะเจาะจง | วิเคราะห์รูปแบบการเช่าตามช่วงเวลา ความแตกต่างระหว่างสมาชิกกับผู้ใช้ทั่วไป และความสัมพันธ์กับสภาพอากาศ |
| **M — Measurable** วัดผลได้ | ตอบคำถามการวิเคราะห์ครบ 14 ข้อ และจัดทำข้อมูลเชิงลึกพร้อมข้อเสนอแนะอย่างน้อย 3 ประเด็น โดยแต่ละประเด็นมีตัวเลขหรือกราฟสนับสนุน |
| **A — Achievable** มีความเป็นไปได้ | ใช้ข้อมูลที่มีอยู่ 731 วัน โดยใช้ Python และ Tableau เตรียมข้อมูล ทำความสะอาดข้อมูล วิเคราะห์ข้อมูล และจัดทำ Data Visualization |
| **R — Relevant** สอดคล้องกับเป้าหมาย | ให้ข้อมูลประกอบการวางแผนเตรียมบริการและกิจกรรมส่งเสริมการใช้จักรยานให้เหมาะกับรูปแบบการใช้บริการ |
 | **T — Time-bound** มีกรอบเวลา | ดำเนินการและนำเสนอผลให้แล้วเสร็จภายใน 4 สัปดาห์นับจากเริ่มโครงงาน |

# 5.	Data Dictionary
| ตัวแปร | ความหมาย | ชนิดข้อมูล | รหัส / หน่วย / รายละเอียด |
|---|---|---|---|
| `instant` | ลำดับระเบียนข้อมูล | จำนวนเต็ม — ตัวระบุ | ใช้ระบุแถวข้อมูล |
| `dteday` | วันที่บันทึกข้อมูล | วันที่ | รูปแบบ `YYYY-MM-DD` |
| `season` | ฤดูกาล | หมวดหมู่ | 1 = ฤดูหนาว, 2 = ฤดูใบไม้ผลิ, 3 = ฤดูร้อน, 4 = ฤดูใบไม้ร่วง |
| `yr` | ปีที่บันทึกข้อมูล | หมวดหมู่ | 0 = ค.ศ. 2011, 1 = ค.ศ. 2012 |
| `mnth` | เดือน | หมวดหมู่ | 1 = มกราคม ถึง 12 = ธันวาคม |
| `holiday` | สถานะวันหยุด | ทวิภาค | 0 = ไม่ใช่วันหยุด, 1 = วันหยุด |
| `weekday` | วันในสัปดาห์ | หมวดหมู่ | รหัส 0–6; ตรวจสอบชื่อวันเทียบกับ `dteday` |
| `workingday` | สถานะวันทำงาน | ทวิภาค | 1 = วันทำงาน, 0 = วันเสาร์–อาทิตย์หรือวันหยุด |
| `weathersit` | ประเภทสภาพอากาศ | หมวดหมู่ | 1 = แจ่มใสหรือมีเมฆบางส่วน; 2 = มีหมอกบางหรือหมอกร่วมกับเมฆ; 3 = ฝนหรือหิมะเล็กน้อย อาจมีพายุฝนฟ้าคะนอง; 4 = ฝนหนักร่วมกับพายุหรือเม็ดน้ำแข็ง หรือหิมะร่วมกับหมอก |
| `temp` | อุณหภูมิที่ปรับสเกลแล้ว | ตัวเลขทศนิยม | ค่า normalized; ตรวจสอบสูตรก่อนแปลงกลับเป็น °C |
| `atemp` | อุณหภูมิที่ร่างกายรู้สึก ซึ่งปรับมาตราส่วนแล้ว | ตัวเลขทศนิยม | ค่า normalized; ตรวจสอบสูตรก่อนแปลงกลับเป็น °C |
| `hum` | ความชื้นสัมพัทธ์ที่ปรับสเกลแล้ว | ตัวเลขทศนิยม | ความชื้นสัมพัทธ์ (%) ÷ 100 |
| `windspeed` | ความเร็วลมที่ปรับมาตราส่วนแล้ว | ตัวเลขทศนิยม | ความเร็วลมเดิม ÷ 67 |
| `casual` | จำนวนครั้งการเช่าของผู้ใช้ทั่วไป | จำนวนเต็ม | ครั้งต่อวัน |
| `registered` | จำนวนครั้งการเช่าของสมาชิก | จำนวนเต็ม | ครั้งต่อวัน |
| `cnt` | จำนวนครั้งการเช่ารวม | จำนวนเต็ม | ครั้งต่อวัน; `cnt = casual + registered` |

* **หลักฐานสนับสนุน:** ผลการวิเคราะห์จากคำถามข้อที่ 1 และแผนภาพแนวโน้มจำนวนครั้งการเช่าจักรยานรายวันตลอดช่วงเวลา 2 ปี

### **ประเด็นที่ 2: พฤติกรรมผู้ใช้ที่แตกต่างชัดเจนระหว่างวันทำงานและวันหยุด (Commuter vs. Leisure Users)**
* **ข้อมูลเชิงลึก (Key Findings):** กลุ่มสมาชิก (Registered) และกลุ่มผู้ใช้ทั่วไป (Casual) มีวัตถุประสงค์ในการใช้งานที่ต่างกันอย่างมาก โดยกลุ่มสมาชิกเน้นการเดินทางเพื่อไปทำงานหรือเรียน สังเกตจากจำนวนครั้งการเช่าเฉลี่ยในวันทำงานที่สูงถึง **3,978 ครั้งต่อวัน** (มากกว่าวันหยุดราว 34%) ในทางตรงกันข้าม กลุ่มผู้ใช้ทั่วไปใช้เพื่อการท่องเที่ยวหรือพักผ่อนเป็นหลัก โดยมีจำนวนครั้งการเช่าเฉลี่ยในวันหยุด **1,371 ครั้งต่อวัน** ซึ่งสูงกว่าวันทำงานปกติถึง **2.3 เท่า**
* **หลักฐานสนับสนุน:** ผลต่างค่าเฉลี่ยพฤติกรรมการเช่าในวันทำงานและวันหยุดจากคำถามข้อที่ 5 และ 11

### **ประเด็นที่ 3: สภาพอากาศที่สบายตัวมีผลต่อการใช้งานมากกว่าอุณหภูมิที่ร้อนจัด (Comfortable Climate over Extreme Heat)**
* **ข้อมูลเชิงลึก (Key Findings):** จำนวนครั้งการเช่าจักรยานไม่ได้แปรผันตรงกับอุณหภูมิที่สูงที่สุดเสมอไป แต่จะมีจำนวนครั้งการเช่าสูงที่สุดในช่วงสภาพอากาศที่เรียกว่า "สบายตัว" คืออุณหภูมิที่ปรับสเกลแล้วอยู่ในช่วง **0.5 ถึงต่ำกว่า 0.6** และมีความชื้นสัมพัทธ์ต่ำเพียง **30% ถึงต่ำกว่า 40%** (จำนวนครั้งการเช่าเฉลี่ยสูงสุดที่ **7,030 ครั้งต่อวัน**) แต่หากอุณหภูมิร้อนจัดในฤดูร้อน หรือมีความชื้นสะสมสูงเกินกว่า 90% (เช่น วันฝนตก) จำนวนครั้งการเช่าจะลดลงเหลือไม่ถึงครึ่งหนึ่งอย่างเห็นได้ชัด
* **หลักฐานสนับสนุน:** การวิเคราะห์แบบหลายตัวแปร (อุณหภูมิและความชื้น) และสหสัมพันธ์รายฤดูกาลในข้อที่ 3, 4, 9, 13 และ 14

#7.	Recommendation

### **ข้อเสนอแนะสำหรับวางแผนการดำเนินงาน (Recommendations for Operational Planning)**

1. **การบริหารจัดการกระจายจักรยานตามความต้องการของแต่ละกลุ่ม (Inventory Optimization):**
   * **วันทำงานปกติ:** ควรจัดเตรียมจักรยานและจุดจอด (Docks) ให้พร้อมในบริเวณสถานีขนส่งสาธารณะ ย่านธุรกิจ และสถานศึกษา เพื่อรองรับกลุ่มสมาชิกหลักที่ใช้เดินทางไปทำงานในชั่วโมงเร่งด่วน
   * **วันหยุดเสาร์-อาทิตย์:** ควรย้ายการสำรองจักรยานไปกระจายอยู่ตามสถานที่ท่องเที่ยว สวนสาธารณะ และเส้นทางเลียบแม่น้ำ เพื่ออำนวยความสะดวกให้กลุ่มผู้ใช้ทั่วไป (Casual Users) ที่เข้ามาใช้งาน

2. **การทำกลยุทธ์การตลาดและแคมเปญโปรโมชันตามสภาพอากาศและฤดูกาล (Season & Weather-based Marketing):**
   * **กระตุ้นการใช้งานช่วง Low Season:** ในช่วงฤดูหนาว หรือวันที่มีความชื้นสูง (เช่น วันฝนตก) ควรออกโปรโมชันลดราคาพิเศษ หรือจัดทำแคมเปญสะสมแต้มคูณสอง เพื่อดึงดูดใจให้ผู้ใช้งานยังคงเลือกใช้จักรยานแทนพาหนะอื่น
   * **โปรโมตช่วงสภาพอากาศดี:** จัดกิจกรรมส่งเสริมการปั่นจักรยานในวันที่พยากรณ์อากาศระบุว่าอุณหภูมิพอดีและความชื้นต่ำกว่า 40% เพื่อผลักดันยอดเช่าให้พุ่งทะลุขีดจำกัด

3. **การแปลงกลุ่มผู้ใช้ทั่วไปให้เป็นสมาชิก (Conversion Strategy for Casual Users):**
   * เนื่องจากกลุ่ม Casual มีจำนวนครั้งการเช่าที่หนาแน่นมากในวันหยุด ควรจัดทำแพ็กเกจสมาชิกประเภทใหม่ เช่น **"Weekend Pass" (สมาชิกเฉพาะวันเสาร์-อาทิตย์)** หรือ **"Leisure Member"** ที่ตอบโจทย์พฤติกรรมรักการท่องเที่ยวปั่นเล่น เพื่อสร้างรายได้ที่สม่ำเสมอ (Recurring Revenue) และดึงกลุ่มผู้ใช้ทั่วไปให้กลายมาเป็นลูกค้าประจำของระบบในระยะยาว

#Q1: จำนวนครั้งการเช่ารายวันในปี 2012 เปลี่ยนแปลงจากปี 2011 เท่าใด?

````python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import StrMethodFormatter

# อ่านข้อมูลและสรุปแยกปี
df = pd.read_csv('/content/drive/MyDrive/Bike_Sharing_Data.csv')
df["dteday"] = pd.to_datetime(df["dteday"], dayfirst=True)
df["year"] = df["dteday"].dt.year

summary = (
    df[df["year"].isin([2011, 2012])]
    .groupby("year")
    .agg(
        days=("dteday", "nunique"),
        total_rentals=("cnt", "sum"),
        daily_average=("cnt", "mean")
    )
)

display(summary.round(2))

# คำนวณการเปลี่ยนแปลง
average = summary["daily_average"]
difference = average.loc[2012] - average.loc[2011]
percent = difference / average.loc[2011] * 100

print(f"ค่าเฉลี่ยเปลี่ยนแปลง {difference:+,.2f} ครั้งต่อวัน ({percent:+.2f}%)")

# สร้างกราฟ
fig, ax = plt.subplots(figsize=(7, 4), dpi=120)

bars = ax.bar(
    average.index.astype(str),
    average.values,
    color=["#FFB7B2", "#B5EAD7"],
    width=0.5
)

ax.bar_label(bars, labels=[f"{v:,.0f}" for v in average], padding=5)

ax.set_title("Average Daily Bike Rentals: 2011 vs 2012", pad=12)
ax.set_ylabel("Average rentals per day")
ax.set_ylim(0, average.max() * 1.15)
ax.yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))

ax.spines[["top", "right"]].set_visible(False)
ax.tick_params(axis="x", length=0)

plt.tight_layout()
plt.show()
````

#Q4: ตัวแปรสภาพอากาศใดมีความสัมพันธ์กับจำนวนครั้งการเช่ารายวันมากที่สุด?

````python
#correlatrion matrix แบบ heatmap
numeric_vec = ['temp', 'atemp', 'hum', 'windspeed',  'cnt']

correlation_matrix = bike[numeric_vec].corr()

# Plot the heatmap of the correlation matrix with 4 decimal places
plt.figure(figsize=(15, 10))
sns.heatmap(correlation_matrix, annot=True, fmt='.4f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Bike Sharing Dataset')
plt.show()
````

````python
#Pair Plot
numeric_vec = ['temp', 'atemp', 'hum', 'windspeed', 'cnt']
sns.pairplot(bike[numeric_vec], diag_kind='kde')
plt.show()
````

#Q9: ความสัมพันธ์ระหว่างความชื้นสัมพัทธ์กับจำนวนครั้งการเช่าแตกต่างกันตามฤดูกาลหรือไม่?

image.png

````python
# หาค่า correlation ระหว่าง cnt และ hum แยกตาม season
for s in sorted(bike['season'].unique()):
    season_df = bike[bike['season'] == s]
    corr_value = season_df['cnt'].corr(season_df['hum'])
    print(f"Season {s}: Correlation between cnt and hum = {corr_value:.4f}")
````

**ข้อสรุปคำถามข้อ 9 :** จากกราฟและค่าสหสัมพันธ์ พบว่าฤดูหนาว (รหัส 1 แทบไม่มีความสัมพันธ์เชิงเส้นระหว่างความชื้นสัมพัทธ์กับจำนวนครั้งการเช่า (r = –0.0393) ส่วนฤดูใบไม้ผลิ (รหัส 2) ฤดูร้อน (รหัส 3) และฤดูใบไม้ร่วง (รหัส 4) มีความสัมพันธ์กันน้อย ในทิศทางตรงกันข้าม  โดยมีค่า r เท่ากับ –0.3199, –0.3203 และ –0.2880 ตามลำดับ

กล่าวคือ เมื่อความชื้นสัมพัทธ์สูงขึ้น จำนวนครั้งการเช่ามีแนวโน้มลดลงเล็กน้อย อย่างไรก็ตาม ที่ระดับความชื้นใกล้เคียงกัน จำนวนครั้งการเช่ายังแตกต่างกันมาก แสเงให้เห็นว่าความชื้นสัมพัทธ์เพียงตัวแปรเดียวอาจไม่เพียงพอที่จะอธิบายความแตกต่างของจำนวนครั้งการเช่า

โครงงานนี้ศึกษาข้อมูลการเช่าจักรยานรายวันของระบบ Capital Bikeshare ระหว่างปี ค.ศ. 2011–2012 จำนวน 731 วัน จาก UC Irvine Machine Learning Repository โดยเริ่มจากการกำหนดวัตถุประสงค์ตามหลัก SMART การตั้งคำถามการวิเคราะห์ และการกำหนดตัวชี้วัด (Metric) และเกณฑ์ความสำเร็จ (KPI) จากนั้นใช้ภาษา Python ในการจัดเตรียมและทำความสะอาดข้อมูล วิเคราะห์ข้อมูลเบื้องต้น (EDA) และวิเคราะห์ความสัมพันธ์ระหว่างตัวแปร ร่วมกับการใช้ Tableau เพื่อนำเสนอผลผ่านกราฟที่เข้าใจง่าย

## 🎯 วัตถุประสงค์เชิงธุรกิจ
วิเคราะห์และค้นหาปัจจัยสำคัญที่มีผลต่อจำนวนการเช่าจักรยานรายวัน โดยระบุแนวโน้ม รูปแบบพฤติกรรมผู้ใช้ และความสัมพันธ์ของสภาพอากาศ เพื่อตอบคำถามทางสถิติและธุรกิจรวม 14 ข้อ และจัดทำแดชบอร์ดสรุปข้อเสนอแนะเชิงปฏิบัติการอย่างมีหลักฐานสนับสนุน

## 🛠 การรวบรวมและเตรียมข้อมูล
ชุดข้อมูลมีขนาด 731 แถว 16 ตัวแปร ครอบคลุมพฤติกรรมของกลุ่มผู้ใช้แบบทั่วไป (Casual) และกลุ่มสมาชิก (Registered)
- **Data Cleaning:** ดำเนินการตรวจสอบค่าว่าง (Missing Value) ข้อมูลซ้ำซ้อน (Duplicate) และความถูกต้องเชิงตรรกะ (`casual` + `registered` = `cnt`) ครบถ้วน 100%
- **Data Transformation:** แปลงข้อมูลประเภทหมวดหมู่ (Season, Month, Weather, Weekday) และแปลงคอลัมน์วันเวลาให้อยู่ในรูปแบบ DateTime ที่ถูกต้องสำหรับวิเคราะห์อนุกรมเวลา (Time Series)

## 📊 การค้นพบที่สำคัญ (Key Findings & Insights)

### 📈 1. อัตราการเติบโตก้าวกระโดด (Year-on-Year Growth)
- ในปี ค.ศ. 2012 มีจำนวนครั้งการเช่าจักรยานรวมเฉลี่ยสูงถึง **5,599.93 ครั้งต่อวัน** เติบโตขึ้นจากปี ค.ศ. 2011 ที่เฉลี่ย **3,405.76 ครั้งต่อวัน** คิดเป็น **+64.43%** แสดงให้เห็นว่าพฤติกรรมการใช้งานที่กลายมาเป็นส่วนหนึ่งในชีวิตประจำวันของผู้คนในเมือง

### 👥 2. พฤติกรรมผู้ใช้แยกกลุ่มที่ชัดเจน (Commuter vs. Leisure)
- **กลุ่มสมาชิก (Registered):** เน้นใช้ในการเดินทางประจำวันทำงานเป็นหลัก มีจำนวนครั้งการเช่าเฉลี่ยสูงถึง **3,978 ครั้งต่อวัน** ในวันทำงานปกติ (มากกว่าวันหยุดพักผ่อน 34%)
- **กลุ่มผู้ใช้ทั่วไป (Casual):** เน้นเช่าเพื่อการพักผ่อนและการท่องเที่ยว ยอดเช่าเฉลี่ยในวันเสาร์-อาทิตย์และวันหยุดสูงถึง **1,371 ครั้งต่อวัน** สูงกว่าวันทำงานปกติถึง **2.3 เท่า**

### 🌦 3. ผลกระทบร่วมกันของอุณหภูมิและความชื้น (Comfortable Climate over Extreme Heat)
- อุณหภูมิสูงไม่ได้ทำให้คนหันมาปั่นจักรยานเพิ่มขึ้นเสมอไป ช่วงสภาพอากาศที่เหมาะสมที่สุด (Sweet Spot) คืออุณหภูมิที่ปรับสเกลแล้วช่วง **0.5 - 0.6** ร่วมกับความชื้นต่ำช่วง **30% - 40%** ซึ่งทำให้จำนวนครั้งการเช่าสูงสุดเฉลี่ย **7,030 ครั้งต่อวัน**
- ในทางกลับกัน หากความชื้นสูงเกินกว่า 90% (เช่น ฝนตก) จำนวนครั้งเช่าเฉลี่ยจะลดลงอย่างชัดเจนเหลือเพียง **3,043 ครั้งต่อวัน**

## 💡 ข้อเสนอแนะเชิงกลยุทธ์ (Recommendations & Action Plan)

1. **การบริหารจำนวนรถและสถานีแบบยืดหยุ่น (Inventory Optimization):**
   - **วันจันทร์-ศุกร์:** มุ่งเน้นการจัดเตรียมระบบจักรยานให้เต็มโควตาบริเวณย่านธุรกิจ (CBD) และสถานีรถไฟเพื่ออำนวยความสะดวกกับกลุ่มเดินทางไปทำงาน (Registered)
   - **วันเสาร์-อาทิตย์:** กระจายรถสำรองไปตามแนวแหล่งท่องเที่ยวธรรมชาติ สวนสาธารณะ และรอบนอกเมืองเพื่อสนับสนุนกลุ่มใช้พักผ่อน (Casual)
2. **แผนการตลาดอิงสภาพอากาศ (Climate-Responsive Marketing):**
   - สร้างโปรโมชันกระตุ้นการใช้งานในวันธรรมดาช่วงฤดูหนาว (Low Season) หรือวันที่มีแนวโน้มสภาพอากาศแปรปรวน เช่น มอบส่วนลดพิเศษแก่กลุ่มผู้เช่าทั่วไป
3. **การแปลงกลุ่ม Casual สู่สมาชิกประจำ (Conversion Strategy):**
   - นำเสนอแผนสมาชิกประเภทใหม่ เช่น **"Weekend Pass"** หรือ **"Leisure Pass"** ที่มีราคาคุ้มค่ากว่าสำหรับผู้ที่รักการปั่นท่องเที่ยวเพื่อเปลี่ยนลูกค้าชั่วคราวให้เป็น Recurring Revenue ในระยะยาว
